```python
df = pd.read_csv("2019.csv")
```

```python
display(df.head())
print("No. of rows",df.shape[0])
print("No. of columns",df.shape[1])
```

```python
print(df.info())
```

```python
display(df.describe())
```

```python
plt.figure(figsize=(10, 6))

sns.histplot(
    df["Score"],
    bins=15,
    kde=True
)

plt.title("Distribution of Happiness Scores - 2019")
plt.xlabel("Happiness Score")
plt.ylabel("Number of Countries")

plt.show()
```

```python
top_10 = df.sort_values(
    by="Score",
    ascending=False
).head(10)

display(top_10[["Country or region", "Score"]])
```

```python
plt.figure(figsize=(12, 6))

sns.barplot(
    data=top_10,
    x="Score",
    y="Country or region"
)

plt.title("Top 10 Happiest Countries - 2019")
plt.xlabel("Happiness Score")
plt.ylabel("Country")

plt.show()
```

```python
correlation = df.select_dtypes(
    include=np.number
).corr()

display(correlation)

plt.figure(figsize=(12, 8))

sns.heatmap(
    correlation,
    annot=True,
    cmap="coolwarm",
    fmt=".2f",
    linewidths=0.5
)

plt.title("Correlation Heatmap - World Happiness Report 2019")

plt.show()
```

```python
plt.figure(figsize=(8, 6))

sns.scatterplot(
    data=df,
    x="Social support",
    y="Score"
)

plt.title("Social Support vs Happiness Score")
plt.xlabel("Social Support")
plt.ylabel("Happiness Score")

plt.show()
```

```python
plt.figure(figsize=(8, 6))

sns.scatterplot(
    data=df,
    x="Generosity",
    y="Score"
)

plt.title("Generosity vs Happiness Score")
plt.xlabel("Generosity")
plt.ylabel("Happiness Score")

plt.show()
```

```python
#outliers
plt.figure(figsize=(10, 5))

sns.boxplot(x=df["Score"])

plt.title("Boxplot of Happiness Scores")
plt.xlabel("Happiness Score")

plt.show()
```

```python
factors = [
    "GDP per capita",
    "Social support",
    "Healthy life expectancy",
    "Freedom to make life choices",
    "Generosity",
    "Perceptions of corruption"
]

for col in factors:
    print("\nFeature:", col)

    display(
        df.nlargest(3, col)[
            ["Country or region", col, "Score"]
        ]
    )

    display(
        df.nsmallest(3, col)[
            ["Country or region", col, "Score"]
        ]
    )
```

```python
print("Average happiness score:", df["Score"].mean())
print("Median happiness score:", df["Score"].median())
print("Minimum happiness score:", df["Score"].min())
print("Maximum happiness score:", df["Score"].max())
```

```python
summary = {
    "Average Happiness Score": df["Score"].mean(),
    "Median Happiness Score": df["Score"].median(),
    "Highest Happiness Score": df["Score"].max(),
    "Lowest Happiness Score": df["Score"].min(),
    "Average GDP per Capita": df["GDP per capita"].mean(),
    "Average Social Support": df["Social support"].mean(),
    "Average Healthy Life Expectancy": df["Healthy life expectancy"].mean()
}

summary_df = pd.DataFrame(
    summary.items(),
    columns=["Metric", "Value"]
)

display(summary_df)
```
