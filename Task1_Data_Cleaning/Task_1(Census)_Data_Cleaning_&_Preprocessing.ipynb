```python
import pandas as pd
import numpy as np

!pip install ucimlrepo
from ucimlrepo import fetch_ucirepo
```

```python
adult = fetch_ucirepo(id=2)

X = adult.data.feature
y = adult.data.targets

df = pd.concat([X,y], axis=1)
print("Dataset loaded successfully")
```

```python
print("First 5 rows:")
display(df.head())
```

```python
print("Dataset shape:",df.shape)
```

```python
print("Dataset Information:",df.info())
```

```python
print("Statistical summary:")
display(df.describe())
```

```python
print("Missing values:",df.isnull().sum())
```

```python
print("Duplicate records:",df.duplicated().sum())
```

```python
df = df.drop_duplicates()

print("Duplicates removed.")

print("Updated shape:", df.shape)
```

```python
numeric_cols = df.select_dtypes(include=np.number).columns
categorial_cols = df.select_dtypes(include="object").columns
```

```python
for col in numeric_cols: df[col].fillna(df[col].median())
```

```python
for col in categorial_cols: df[col]=df[col].fillna("Unknown")
```

```python
for col in categorial_cols: df[col]=df[col].astype(str).str.strip()
```

```python
df["age"]=df["age"].astype(int)
df["hours-per-week"]=(df["hours-per-week"].astype(int))
```

```python
#verify
print("\nMissing Value",df.isnull().sum().sum())
print("\nDuplicate Records",df.duplicated().sum())
print("\nShape",df.shape)
print("\nInfo",df.info)
```

```python
df.to_csv(
    "cleaned_adult_income.csv",
    index=False
)

print("Cleaned dataset exported successfully!")
```
